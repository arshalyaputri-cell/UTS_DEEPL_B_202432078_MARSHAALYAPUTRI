```python
#202432078_Marsha Alya Putri
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">2. PRA-PEMROSESAN DATA (PREPROCESSING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat dataset, membersihkan data, melakukan normalisasi / tokenisasi sekuens, serta membagi data latih dan data uji.</p>
</div>

```python
#202432078_Marsha Alya Putri
# 1. Muat dataset (pastikan PJME_hourly.csv berada satu folder dengan notebook)
df = pd.read_csv('PJME_hourly.csv', parse_dates=['Datetime'])
print("Ukuran data awal :", df.shape)
print(df.head())

# 2. Pembersihan data: urutkan waktu, buang timestamp ganda, isi jam yang hilang
df = df.sort_values('Datetime').drop_duplicates(subset='Datetime').set_index('Datetime')
df = df.asfreq('h').interpolate(method='linear')
print("\nUkuran data bersih :", df.shape)
print("Jumlah missing value :", df['PJME_MW'].isnull().sum())
print(df.describe())

# 3. Visualisasi data mentah
plt.figure(figsize=(14, 4))
plt.plot(df.index, df['PJME_MW'], color='#114F6A', linewidth=0.5)
plt.title('Konsumsi Listrik PJME per Jam (2002-2018)')
plt.xlabel('Waktu')
plt.ylabel('Konsumsi (MW)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()

# 4. Split data secara kronologis (80% latih, 20% uji)
data = df['PJME_MW'].values.astype(np.float32).reshape(-1, 1)
train_size = int(len(data) * 0.8)
train_raw = data[:train_size]
test_raw = data[train_size:]

# 5. Normalisasi Min-Max (parameter dihitung hanya dari data latih)
data_min = train_raw.min()
data_max = train_raw.max()
train_scaled = (train_raw - data_min) / (data_max - data_min)
test_scaled = (test_raw - data_min) / (data_max - data_min)

# 6. Sliding window: 24 jam terakhir -> prediksi 1 jam berikutnya
seq_length = 24

def create_sequences(data, seq_length):
    xs, ys = [], []
    for i in range(len(data) - seq_length):
        xs.append(data[i:(i + seq_length)])
        ys.append(data[i + seq_length])
    return torch.tensor(np.array(xs), dtype=torch.float32), torch.tensor(np.array(ys), dtype=torch.float32)

# data uji diberi 24 jam terakhir data latih sebagai konteks awal
test_input = np.concatenate([train_scaled[-seq_length:], test_scaled])

#202432078_Marsha Alya Putri
# 7. Konversi ke PyTorch Tensor
X_train, y_train = create_sequences(train_scaled, seq_length)
X_test, y_test = create_sequences(test_input, seq_length)
print("\nX_train:", X_train.shape, "| y_train:", y_train.shape)
print("X_test :", X_test.shape, "| y_test :", y_test.shape)

# 8. DataLoader untuk training per batch
batch_size = 256
train_loader = DataLoader(TensorDataset(X_train, y_train), batch_size=batch_size, shuffle=True)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">3. PEMODELAN ARSITEKTUR JARINGAN (RNN / LSTM)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Mendefinisikan kelas arsitektur model sekuensial menggunakan modul torch.nn sesuai algoritma yang dipilih.</p>
</div>

```python
#202432078_Marsha Alya Putri
class SequenceModel(nn.Module):
    def __init__(self, input_size=1, hidden_size=64, output_size=1):
        super(SequenceModel, self).__init__()
        self.lstm = nn.LSTM(input_size, hidden_size, batch_first=True)
        self.linear = nn.Linear(hidden_size, output_size)

    def forward(self, x):
        lstm_out, (hn, cn) = self.lstm(x)
        pred = self.linear(lstm_out[:, -1, :])
        return pred
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">4. INISIASI HYPERPARAMETER & PROSES PELATIHAN (TRAINING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menentukan konfigurasi hyperparameter (learning rate, epochs, loss function, optimizer) dan menjalankan siklus pelatihan model.</p>
</div>

```python
#202432078_Marsha Alya Putri
# 1. Inisiasi hyperparameter
epochs = 20
learning_rate = 0.001
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print("Perangkat yang dipakai:", device)

# 2. Inisiasi model, loss function, dan optimizer
model = SequenceModel().to(device)
criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=learning_rate)

# 3. Simpan riwayat loss
train_losses = []
test_losses = []
X_test_dev, y_test_dev = X_test.to(device), y_test.to(device)

# 4. Training loop
print("=== Melatih Model LSTM ===")
for epoch in range(epochs):
    model.train()
    running_loss = 0.0
    for xb, yb in train_loader:
        xb, yb = xb.to(device), yb.to(device)
        optimizer.zero_grad()
        y_pred = model(xb)
        loss = criterion(y_pred, yb)
        loss.backward()
        optimizer.step()
        running_loss += loss.item() * xb.size(0)

    epoch_loss = running_loss / len(train_loader.dataset)
    train_losses.append(epoch_loss)

    model.eval()
    with torch.no_grad():
        test_loss = criterion(model(X_test_dev), y_test_dev).item()
    test_losses.append(test_loss)

    if (epoch + 1) % 2 == 0:
        print(f"Epoch {epoch+1}/{epochs} - Train Loss: {epoch_loss:.6f} - Test Loss: {test_loss:.6f}")
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">5. VISUALISASI LOSS TRAINING & PERBANDINGAN PREDIKSI VS AKTUAL</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menampilkan grafik konvergensi nilai Loss terhadap Epoch dan grafik evaluasi perbandingan hasil prediksi model terhadap data aktual.</p>
</div>

```python
#202432078_Marsha Alya Putri
# 1. Grafik penurunan loss training terhadap epoch
plt.figure(figsize=(10, 4))
plt.plot(train_losses, color='#114F6A', linewidth=2, label='Training Loss')
plt.plot(test_losses, color='#1FA4B1', linewidth=2, linestyle='--', label='Test Loss')
plt.title('Grafik Penurunan Loss Selama Pelatihan')
plt.xlabel('Epoch')
plt.ylabel('Loss (MSE)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()

# 2. Prediksi pada data uji lalu kembalikan ke skala MW
model.eval()
with torch.no_grad():
    pred_scaled = model(X_test_dev).cpu().numpy()
y_pred_mw = pred_scaled * (data_max - data_min) + data_min
y_true_mw = y_test.numpy() * (data_max - data_min) + data_min

# 3. Metrik evaluasi
mae = np.mean(np.abs(y_true_mw - y_pred_mw))
rmse = np.sqrt(np.mean((y_true_mw - y_pred_mw) ** 2))
mape = np.mean(np.abs((y_true_mw - y_pred_mw) / y_true_mw)) * 100
print(f"MAE  : {mae:.2f} MW")
print(f"RMSE : {rmse:.2f} MW")
print(f"MAPE : {mape:.2f} %")

# 4. Grafik Data Aktual vs Prediksi (7 hari pertama data uji = 168 jam)
plt.figure(figsize=(12, 5))
plt.plot(y_true_mw[:168], color='#114F6A', linewidth=2, label='Data Aktual')
plt.plot(y_pred_mw[:168], color='#1FA4B1', linewidth=2, linestyle='--', label='Prediksi LSTM')
plt.title('Perbandingan Data Aktual vs Prediksi LSTM (168 Jam Pertama Data Uji)')
plt.xlabel('Jam ke-')
plt.ylabel('Konsumsi (MW)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()

# 5. Grafik seluruh data uji
plt.figure(figsize=(14, 5))
plt.plot(y_true_mw, color='#114F6A', alpha=0.6, linewidth=0.6, label='Data Aktual')
plt.plot(y_pred_mw, color='#1FA4B1', alpha=0.8, linewidth=0.6, label='Prediksi LSTM')
plt.title('Perbandingan Data Aktual vs Prediksi LSTM (Seluruh Data Uji)')
plt.xlabel('Jam ke-')
plt.ylabel('Konsumsi (MW)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()
```
